### Создайте одномерный массив со значениями от 10 до 49 и разверните его (первый элемент становится последним). Найдите в нем все четные элементы.

```python
massiv = np.arange(10,50)
new_massiv = massiv[::-1]
a = new_massiv[new_massiv%2==0]
print(a)
```

### Создайте двумерный массив 3x3 со значениями от 0 до 8

```python
massiv = np.arange(0,9)
new_massiv= massiv.reshape(3,3)
print(new_massiv)
```

### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
rng = np.random.default_rng()
massiv = rng.random((4, 3, 2))
min_value = massiv.min()
max_value = massiv.max()
print (massiv)
print("Минимум:", min_value)
print("Максимум:", max_value)


```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение.

```python
rng = np.random.default_rng()
a = rng.integers(low=1, high=15, size=(6,4))
b = rng.integers(low=1, high=15, size=(4,3))
c = a@b
print("Массив a:\n", a)
print("\nМассив b:\n", b)
print("\nМассив c:\n", c)
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
rng = np.random.default_rng()
massiv = rng.integers(0, 101, size=(7, 7))
new_massiv = (massiv-np.mean(massiv))/np.std(massiv)
print (massiv)
print (f'Среднее отклонение:{np.mean(massiv):.2f}')
print (f'Страндартное отклонение:{np.std(massiv):.2f}')
print("\nНормализованный массив (округленный):")
print (new_massiv.round(2))
```
