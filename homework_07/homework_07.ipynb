### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
rng = np.random.default_rng()
massiv = rng.random((4, 3, 2))
min_value = massiv.min()
max_value = massiv.max()
print (massiv)
print("Минимум:", min_value)
print("Максимум:", max_value)


```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение.

```python
rng = np.random.default_rng()
a = rng.integers(low=1, high=15, size=(6,4))
b = rng.integers(low=1, high=15, size=(4,3))
c = a@b
print("Массив a:\n", a)
print("\nМассив b:\n", b)
print("\nМассив c:\n", c)
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
rng = np.random.default_rng()
massiv = rng.integers(0, 101, size=(7, 7))
new_massiv = (massiv-np.mean(massiv))/np.std(massiv)
print (massiv)
print (f'Среднее отклонение:{np.mean(massiv):.2f}')
print (f'Страндартное отклонение:{np.std(massiv):.2f}')
print("\nНормализованный массив (округленный):")
print (new_massiv.round(2))
```

### Найдите количество курящих людей

```python
x = (tips["smoker"]=="Yes").sum()
print(x)
```

### Узнайте какой средний 'total_bill' в зависимости от 'day'

```python
result=tips.groupby('day')['total_bill'].mean().round(2)
print (result)
```

### Отберите строчки с 'total_bill' больше медианы и узнайте какой средний 'tip' в зависимости от 'sex'

```python
x=tips.total_bill.median()
new_table=tips[tips.total_bill>x]
b= new_table.groupby("sex")['tip'].mean().round(2)
print (f'Медиана: {x}')
print(new_table)
print(f'Зависимость:{b}')
```

### Преобразуйте признак 'smoker' в бинарный (0-No, 1-Yes)

# III. Visualization

### Постройте гистограмму распределения признака 'total_bill'

```python
sns.histplot(data=tips, x='total_bill', kde=True)
fig = px.histogram(tips, x="total_bill")
plt.show()
fig.show()
```

### Постройте scatterplot, представляющий взаимосвязь между признаками 'total_bill' и 'tip'

```python
sns.scatterplot(data=tips, x='total_bill', y='tip')
print('Связь между счетом и чаевыми')
plt.show()
```

### Постройте pairplot

```python
sns.pairplot(tips, hue='sex', palette='husl')
plt.show()
```

### Постройте график взаимосвязи между признаками 'total_bill' и 'day'

```python
sns.scatterplot(data=tips, x='total_bill', y='day')
print('Связь между счетом и днём')
plt.show()
```

### Постройте две гистограммы распределения признака 'tip' в зависимости от категорий 'time'

```python
sns.displot(data=tips, x='tip', col='time', kde=True)
fig = px.histogram(tips, x="tip", facet_col="time")
plt.show()
fig.show()
```

### Постройте два графика scatterplot, представляющих взаимосвязь между признаками 'total_bill' и 'tip' один для Male, другой для Female и раскрасьте точки в зависимоти от признака 'smoker'

```python
sns.relplot(
    data=tips,
    x="total_bill", y="tip", col="sex",
    hue="smoker")
plt.show()
```
